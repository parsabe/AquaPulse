```python
from ultralytics import YOLO

# 1. Path to the trained YOLOv8 model (e.g., 'yolov8n.pt' for the nano version)
model = YOLO("yolov8n.pt")

# 2. track the video using the trained model and BoTSORT tracker
results = model.track(
    source="btest.mp4",     # source to video file
    tracker="botsort.yaml", # tracker algorithm selection
    conf=0.25,              # object detection confidence threshold
    iou=0.5,                # IOU thresholds
    show=True,              # display results in real-time
    save=True               # save results to 'runs/detect/track' directory
)
```
