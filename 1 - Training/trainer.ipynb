```python
from ultralytics import YOLO


def main():
    model = YOLO("yolov8m.pt")  # pretrained weights for transfer learning

    # --- 2. Training ---
    results = model.train(
        data="data.yaml",
        epochs=150,            # 
        imgsz=640,             # 
        batch=16,              # adjust based on GPU memory
        patience=30,           # stop if no improvement for 30 epochs
        device=0,              # 0 for GPU, 'cpu' for CPU
        workers=8,
        optimizer="AdamW",
        lr0=0.001,
        cos_lr=True,           # cosine learning rate scheduler -> more stabile convergence
        project="fish_detection",
        name="yolov8m_fish_v1",

        # --- Augmentation ---
        hsv_h=0.015,
        hsv_s=0.7,
        hsv_v=0.4,
        degrees=10.0,          # slight rotation
        translate=0.1,
        scale=0.5,
        shear=2.0,
        flipud=0.3,            # slight vertical flip
        fliplr=0.5,
        mosaic=1.0,
        mixup=0.1,
        close_mosaic=10,       # close mosaic for last 10 epochs to focus on harder examples

        save=True,
        save_period=10,
        val=True,
        plots=True,
        seed=42,
        deterministic=True,
        verbose=True,
    )

    # --- 3. Evaluate best model in the test set ---
    best_model = YOLO("fish_detection/yolov8m_fish_v1/weights/best.pt")
    metrics = best_model.val(data="data.yaml", split="test")
    print("Test sonuçları:")
    print(f"  mAP50:    {metrics.box.map50:.4f}")
    print(f"  mAP50-95: {metrics.box.map:.4f}")
    print(f"  Precision:{metrics.box.mp:.4f}")
    print(f"  Recall:   {metrics.box.mr:.4f}")


if __name__ == "__main__":
    main()
```
